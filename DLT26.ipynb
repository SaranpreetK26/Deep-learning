```python
sentence = "I love deep learning."
tokens = sentence.split()

vocabulary={
    word: index
    for index, word in enumerate(tokens, start=1)
}

token_ids=[
    vocabulary[word]
    for word in tokens
]
print("Original sentence:", sentence)
print("Tokens:", tokens)
print("Vocabulary:", vocabulary)
print("Token IDs:", token_ids)
```

```python
#attention

import tensorflow as tf
#Step 1: Define three query vectors
Q=tf.constant([
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
])

#Step 2: Define three key vectors
K=tf.constant([
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
])

#Step 3: Define V
V=tf.constant([
    [1.0, 0.0],
    [0.0, 10.0],
    [5.0, 5.0]
])

#Step 4: Calculate query key score
score=tf.matmul(Q, K, transpose_b=True)

#Step 5: Scale the scores
key_dimension=tf.cast(
    tf.shape(K)[-1],
    tf.float32
)
scaled_score=score/tf.sqrt(key_dimension)

#Step 6: Convert score into attention weights
weights=tf.nn.softmax(scaled_score, axis=-1)

#Step 7: Calculate the attention output
output=tf.matmul(weights, V)

print("Scores:\n", score.numpy())
print("Scaled scores:\n", scaled_score.numpy())
print("Attention weights:\n", weights.numpy())
print("Output:\n", output.numpy())
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

class Transformation(layers.Layer):
  def __init__(self, d_model, num_heads, ff_dim, dropout=0.1):

    super().__init__()

    if d_model % num_heads != 0:
      raise ValueError(
          f"d_model = {d_model} should be divisible by num_heads = {num_heads}"
      )

    # Multi head self attention
    self.attention = layers.MultiHeadAttention(
        num_heads=num_heads,
        key_dim=d_model,
        dropout=dropout
    )
```

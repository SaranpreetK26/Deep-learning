```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import (load_img,img_to_array,ImageDataGenerator)
```

```python
image = load_img("/content/RC.jpg")
image_array =img_to_array (image)
```

```python
# Add batch dimesnion
image_array = np.expand_dims (image_array, axis=0)
print ("Image shape", image_array.shape)
```

```python
from textwrap import fill
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode="nearest"
)
```

```python
augmentation_generator= datagen.flow(image_array,batch_size=1)
```

```python
# Define output folder and create if it doesn't exist
from tensorflow.keras.preprocessing.image import array_to_img
import os
output_folder = "/content/augmented_images"
os.makedirs(output_folder, exist_ok=True)

# Create augmentation generator
augmentation_generator = datagen.flow(
    image_array,
    batch_size=1
)

# Generate and save 9 images
for i in range(100):

    augmented_image = next(augmentation_generator)[0]

    # Convert array to image
    img = array_to_img(augmented_image)

    # Save image
    image_path = f"{output_folder}/augmented_image_{i+1}.jpg"

    img.save(image_path)

    print("Saved:", image_path)
```

```python
import shutil

shutil.make_archive(
    "/content/augmented_images",
    "zip",
    output_folder
)

print("ZIP file created successfully!")
```
